### Simple If, elif

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")  # This runs because 85 >= 80 is True
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```

### If with logical condition

```python
has_ticket = True
is_vip = False

if has_ticket or is_vip:
    print("Welcome to the event!")  # Runs because has_ticket is True
```

```python
# Example values
exam_score = 85
attendance = 70

if exam_score > 80 and attendance > 75:
    print("Result: Admitted")
elif exam_score > 80 and (attendance > 65 and attendance < 75):
    print("Result: Conditionally passed")
else:
    print("Result: Failed")
```

```python
# Example values
exam_score = 75
attendance = 95

# 1. Check for Full Admission
# Criteria: Exam score must be over 80 AND attendance must be over 75
if exam_score > 80 and attendance > 75:
    print("Result: Admitted")

# 2. Check for Conditional Passing
# This uses two distinct pathways separated by the 'or' operator:
#   - Pathway A: Score > 80, but attendance falls between 65 and 75
#   - Pathway B: Attendance > 90, but score falls between 70 and 80
elif (exam_score > 80 and (attendance > 65 and attendance < 75)) or \
     (attendance > 90 and (exam_score > 70 and exam_score < 80)):
    print("Result: Conditionally passed")

# 3. Fallback Condition
# If the student does not meet any of the criteria above, they fail
else:
    print("Result: Failed")
```

### match and case 

```python
# match and case example
status_code = 200

match status_code:
    # Match a single value
    case 200:
        print("Status: Success (OK)")
        
    case 201:
        print("Status: Created successfully")
        
    # Match multiple alternative values using the pipe (|) operator
    case 400 | 422:
        print("Status: Client Error (Bad Request or Unprocessable Entity)")
        
    case 404:
        print("Status: Not Found")
        
    # Match server errors
    case 500 | 502:
        print("Status: Server Error (Internal Error or Bad Gateway)")
        
    # Default case (wildcard) if none of the above match
    case _:
        print("Status: Unknown or unhandled status code")
```

### Loop Control Statements

```python
# ==========================================
# Example A: Using 'break' to stop early
# Completely exits (stops) the loop immediately.
# ==========================================
print("--- Searching for a target item ---")
fruits = ["apple", "banana", "cherry", "date"]

for fruit in fruits:
    if fruit == "cherry":
        print("Found cherry! Stopping search.")
        break  # Exits the loop entirely
    print(f"Checked: {fruit}")
```

```python
# ==========================================
# Example B: Using 'continue' to skip an item
# Skips the rest of the current iteration and jumps straight to the next iteration.
# ==========================================
print("\n--- Printing only odd numbers ---")
for num in range(1, 6):
    if num % 2 == 0:
        continue  # Skips even numbers and moves to the next loop
    print(num)
```

```python
# ==========================================
# Example C: Prime Number Search using 'for-else'
# The else block runs only if the loop finishes normally 
# (i.e., it completed all iterations without being prematurely stopped by a break statement)
# ==========================================
number_to_check = 7
is_prime = True

for i in range(2, number_to_check):
    if number_to_check % i == 0:
        print(f"{number_to_check} is not a prime number.")
        break
else:
    # This block executes ONLY if the loop finishes without hitting 'break'
    print(f"{number_to_check} is a prime number!")
```

### List Comprehensions

```python
# Standard way: Get all even numbers squared from 1 to 10
even_squares = []
for num in range(1, 11):
    if num % 2 == 0:
        even_squares.append(num ** 2)

print(even_squares)
# Pythonic List Comprehension way: [expression FOR item IN iterable IF condition]
even_squares_compact = [num ** 2 for num in range(1, 11) if num % 2 == 0]

print(even_squares_compact)  # Output: [4, 16, 36, 64, 100]
```

### Advanced Example - Palindrome Checker
A palindrome is a word that reads the same forwards and backwards, like "racecar".
This checker uses two pointers—one starting at the first letter and one at the last.
A `while` loop compares the letters at both positions and moves inward step by step.
If any mismatched letters are found, the loop stops and flags it as not a palindrome.
If the pointers meet successfully in the middle, it confirms the word is a valid palindrome!

```python
# ==========================================
# Palindrome Checker using a Two-Pointer While Loop
# ==========================================

word = "racecar"

# Initialize pointers
left = 0
right = len(word) - 1
is_palindrome = True

# Loop continues as long as the left pointer hasn't crossed the right pointer
while left < right:
    # If characters at both pointers do not match, it's not a palindrome
    if word[left] != word[right]:
        is_palindrome = False
        break  # Exit the loop early since we found a mismatch
    
    # Move pointers inward
    left += 1
    right -= 1

if is_palindrome:
    print(f"'{word}' is a palindrome!")
else:
    print(f"'{word}' is not a palindrome.")
```

```python
# ==========================================
# Example 1: Calculate Rectangle Area
# ==========================================

def calculate_area(length, width):
    """Calculates the area of a rectangle."""
    area = length * width
    return area

# Calling the function and storing the result
result = calculate_area(length=5, width=3)
print(f"The area is: {result}")  # Output: The area is: 15
```

```python
# ==========================================
# Example 2: Personalized Greeting
# ==========================================

def greet_user(name, greeting="Hello"):
    """Greets a user with a customizable greeting."""
    return f"{greeting}, {name}!"

# Using the default greeting
print(greet_user(name = "Alice"))          # Output: Hello, Alice!

# Overriding the default greeting
print(greet_user(name = "Bob", greeting = "Welcome"))  # Output: Welcome, Bob!
```

```python
# ==========================================
# Example 3: Min and Max Calculator
# An example of a function returning multiple values
# ==========================================

def get_min_and_max(numbers):
    """Returns both the smallest and largest number from a list."""
    return min(numbers), max(numbers)

# Unpacking the returned values into two separate variables
scores = [45, 89, 23, 99, 67]
lowest, highest = get_min_and_max(scores)

print(f"Lowest score: {lowest}")   # Output: Lowest score: 23
print(f"Highest score: {highest}") # Output: Highest score: 99
```

```python
# ==========================================
# Example 4:Function with Parameter and Return Type Hints
# ==========================================

# Parameter Annotations (name: type): Placed directly after the parameter name inside the parentheses.
# Return Type Annotations (-> type): Placed right before the colon that ends the function definition header.
# Optional Documentation: Modern code editors (like VS Code or PyCharm) use these hints to warn you if you accidentally 
# pass the wrong data type (like passing a string instead of a float).

def calculate_total(price: float, tax_rate: float) -> float:
    """
    Calculates the total price including tax.
    
    :param price: The base price of the item (float)
    :param tax_rate: The tax percentage as a decimal (float)
    :return: The final calculated total (float)
    """
    total = price + (price * tax_rate)
    return total

# Calling the function with correct types
final_price = calculate_total(price = 49.99, tax_rate = 0.08)
print(f"Total Price: ${final_price:.2f}")  # Output: Total Price: $53.99
```

### Lambda Function
A **lambda function** is a small, anonymous (nameless) function in Python. Unlike regular functions defined with `def`, lambda functions are written in a single line using the `lambda` keyword and can take any number of arguments, but can only have **one expression**.

```python
# Regular function
def add(a, b):
    return a + b

# Lambda equivalent assigned to a variable
add_lambda = lambda a, b: a + b

print(add_lambda(3, 5))  # Output: 8
```

## **Sorting**

### Sorting a list

```python
# ==========================================
# Example 1: Sorting Numbers and Strings In-Place
# The original list gets updated
# ==========================================

numbers = [5, 2, 9, 1, 5, 6]
numbers.sort()
print("Sorted numbers:", numbers)  # Output: [1, 2, 5, 5, 6, 9]

# Sorting strings alphabetically
fruits = ["banana", "apple", "cherry", "date"]
fruits.sort()
print("Sorted fruits:", fruits)   # Output: ['apple', 'banana', 'cherry', 'date']
```

```python
# ==========================================
# Example 2: Using sorted()
# A new list is created
# ==========================================

original_scores = [85, 42, 95, 60]

# Creates a new list, leaving original_scores unchanged
sorted_scores = sorted(original_scores)

print("Original:", original_scores)  # Output: [85, 42, 95, 60]
print("Sorted:", sorted_scores)      # Output: [42, 60, 85, 95]
```

```python
# ==========================================
# Example 3: Descending Order
# ==========================================

prices = [19.99, 5.49, 45.00, 12.50]

# Using sorted() with reverse=True
expensive_first = sorted(prices, reverse=True)

print(expensive_first)  # Output: [45.00, 19.99, 12.50, 5.49]
```

```python
# ==========================================
# Example 4: Sorting Strings by Length
# You can customize how items are compared by passing a function to the key parameter.
# ==========================================

words = ["python", "is", "an", "awesome", "language"]

# Sort by the length of each word instead of alphabetically
words.sort(key=len)

print(words)  # Output: ['is', 'an', 'python', 'awesome', 'language']
```

```python
# ==========================================
# Example 5: Sorting List of Tuples by Column (Grade)
# You can customize how items are compared by passing a lambda function to the key parameter.
# ==========================================
# A list of student tuples: (Name, Grade)
students = [("Alice", 85), ("Bob", 92), ("Charlie", 78)]

# Sort the students by their grade (index 1) using a lambda function
sorted_students = sorted(students, key=lambda student: student[1])

print(sorted_students)
```

## **Exception Handling**

```python
def divide_numbers(a, b):
    """Safely divides two numbers with comprehensive exception handling."""
    try:
        # Code that might raise an exception
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero!")
        return None
    except TypeError:
        print("Error: Both inputs must be numbers (int or float)!")
        return None
    else:
        # Runs ONLY if the try block succeeds without errors
        print("Division calculated successfully.")
        return result
    finally:
        # Always runs, regardless of whether an exception occurred or not
        print("--- Execution Finished ---\n")
```

```python
print(divide_numbers(10, 2))    # Successful division
```

```python
print(divide_numbers(10, 0))    # Triggers ZeroDivisionError
```

```python
print(divide_numbers(10, "2"))  # Triggers TypeError
```

### Best Practices for Exception Handling

* **Catch Specific Exceptions**: Always catch the precise exception type (e.g., `ValueError`, `KeyError`, `FileNotFoundError`) rather than using a bare `except:` clause. Catching everything blindly can mask underlying bugs and make debugging very difficult.
* **Keep `try` Blocks Small**: Wrap only the specific lines of code that are likely to fail inside the `try` block. This ensures you know exactly which line raised the error.
* **Use the `else` Block**: Place code that should execute only if the `try` block ran successfully inside the `else` block. This keeps your `try` block clean and strictly focused on error-prone code.
* **Use `finally` for Cleanup**: Use the `finally` block to release external resources (such as closing database connections, open files, or network sockets) because it guarantees execution no matter what. (Alternatively, use context managers like `with`).
* **Avoid Silent Failures (`pass`)**: Never leave an `except` block completely empty or use `except: pass` unless there is a very deliberate, documented reason. Silencing errors hides problems that may cause erratic behavior later in your program.

### JSON to Dictionary and Vice Versa

```python
# A Python dictionary
student = {
    "name": "Alice",
    "age": 22,
    "is_student": True,
    "courses": ["Math", "Physics"]
}

# Convert Python dictionary to a JSON string
json_string = json.dumps(student, indent=4)

print(json_string)
```

```python
# A JSON string (note that boolean values use lowercase `true` in JSON)
py_dict = json.loads(json_string)
print(py_dict)
```

### File Operations with JSON

```python
# Open a file in write mode and dump the data
with open("data.json", "w") as file:
    json.dump(json_string, file, indent=4)
```

```python
# Open the file in read mode and load the data
with open("data.json", "r") as file:
    loaded_data = json.load(file)

print(loaded_data)
dict_data = json.loads(loaded_data)
print(dict_data)
```
