##  **Person Class**

The `Person` class serves as the root foundation of our object-oriented hierarchy. It encapsulates core, universal attributes and behaviors that are shared by every individual in the system, regardless of whether they are a regular employee, a manager, or any future role we might introduce.

### Design Principles Demonstrated
* **Encapsulation:** Groups related data (`name`, `age`, `gender`) into a single unit.
* **Reusability:** Prevents code duplication by centralizing personal details so child classes can inherit them automatically.

### Attributes
* **`name`** (`str`): The full name of the person.
* **`age`** (`int`): The age of the person in years.
* **`gender`** (`str`): The gender identity of the person.

### Methods
* **`__init__(name, age, gender)`**: Constructor initializing personal attributes.
* **`get_details()`**: Returns a base dictionary of personal attributes (designed to be extended by child classes).

---


```python
# Level 1: Base Class

class Person:
    """A base class representing a generic person with personal details."""

    def __init__(self, name: str, age: int, gender: str):
        """Initializes the Person attributes: name, age, and gender."""
        self.name = name
        self.age = age
        self.gender = gender

    def get_details(self) -> dict:
        """Returns a dictionary containing basic personal details (JSON object)."""
        return {"name": self.name, 
                "age": self.age, 
                "gender": self.gender}
```

## **Employee Class**
**Role:** Derived (Child) Class inheriting from `Person`

The `Employee` class represents a specialized type of `Person`—specifically, someone employed by the organization. It inherits all attributes and methods from the `Person` base class via single inheritance and extends them with professional attributes (`department`, `salary`) and compensation logic.

### Design Principles Demonstrated
* **Inheritance:** Automatically gains `name`, `age`, and `gender` from `Person` without rewriting initialization code (using `super().__init__()`).
* **Extension:** Adds employee-specific data fields and behavior (`calculate_pay`).
* **Method Chaining:** Uses `super().get_details()` to build upon the parent's dictionary rather than replacing it entirely.

### Attributes
* *(Inherited from `Person`)*: `name`, `age`, `gender`
* **`department`** (`str`): The corporate department the employee works in.
* **`salary`** (`float`): The base monetary compensation of the employee.

### Methods
* **`__init__(name, age, gender, department, salary)`**: Initializes both inherited `Person` data and `Employee`-specific data.
* **`calculate_pay()`**: Returns the standard base salary.
* **`get_details()`**: Extends `Person`'s dictionary to include role, department, salary, and calculated pay.

```python
# Level 2: Derived Class inheriting from Person
class Employee(Person):
    """A derived class representing an employee, inheriting from Person."""

    def __init__(
        self, name: str, age: int, gender: str, department: str, salary: float
    ):
        """Initializes both Person attributes (via super()) and Employee-specific attributes."""
        super().__init__(name, age, gender)
        self.department = department
        self.salary = salary

    def calculate_pay(self) -> float:
        """Calculates and returns the standard base salary for a regular employee."""
        return self.salary

    def get_details(self) -> dict:
        """Extends Person's get_details dictionary to include department and pay info."""
        # Retrieve the base person dictionary using super()
        details = super().get_details()
        # Add employee-specific keys and values
        details.update(
            {
                "role": "Employee",
                "department": self.department,
                "salary": self.salary,
                "total_pay": self.calculate_pay(),
            }
        )
        return details
```

## **Manager Class**
**Role:** Multi-Level Derived Class inheriting from `Employee`

The `Manager` class represents an advanced, specialized tier of employee. It demonstrates multi-level inheritance: `Manager` inherits from `Employee`, which in turn inherits from `Person` ($Manager \rightarrow Employee \rightarrow Person$).

This class overrides behavior defined in its parent class (`Employee`) to accommodate unique managerial rules, such as team size tracking and a mandatory 10% performance bonus added on top of the base salary.

#### Design Principles Demonstrated
* **Multi-Level Inheritance:** Accesses attributes and methods across a 3-tier hierarchical chain seamlessly.
* **Method Overriding (Polymorphism):** Redefines `calculate_pay()` and `get_details()` to implement specialized bonus and formatting logic tailored specifically for managers.
* **Code Delegation (`super()`):** Reuses parent calculation logic before applying managerial adjustments.

#### Attributes
* *(Inherited from `Person`)*: `name`, `age`, `gender`
* *(Inherited from `Employee`)*: `department`, `salary`
* **`team_size`** (`int`): The number of direct reports managed by this individual.
* **`bonus`** (`int`): 10% of the base salary

#### Methods
* **`__init__(name, age, gender, department, salary, team_size)`**: Initializes the full ancestral chain (`Employee`/`Person`) plus team size.
* **`calculate_pay()`**: Overrides `Employee.calculate_pay()` to add a 10% manager bonus.
* **`get_details()`**: Overrides `Employee.get_details()` to update role, team size, and total pay.

```python
# Level 3: Derived Class inheriting from Employee (Multi-level Inheritance)
class Manager(Employee):
    """A multi-level derived class representing a manager, inheriting from Employee."""

    def __init__(
        self,
        name: str,
        age: int,
        gender: str,
        department: str,
        salary: float,
        team_size: int,
    ):
        """Initializes Employee and Person attributes (via super()) plus manager team size."""
        super().__init__(name, age, gender, department, salary)
        self.team_size = team_size
        self.bonus = salary*0.1 # Calculate 10% manager bonus

    # OVERRIDING calculate_pay from Employee
    def calculate_pay(self) -> float:
        """Overrides Employee.calculate_pay to add a 10% manager bonus on top of base salary."""
        base_pay = super().calculate_pay()
        return base_pay + self.bonus

    # OVERRIDING get_details from Employee
    def get_details(self) -> dict:
        """Extends Employee's get_details dictionary with manager role, team size, and updated bonus pay."""
        # Retrieve employee details dictionary from the parent class
        details = super().get_details()
        # Update values and add manager-specific fields
        details["role"] = "Manager"
        details["team_size"] = self.team_size
        details["total_pay"] = self.calculate_pay()  # Uses overridden method
        details["bonus"] = self.bonus
        return details
```

```python
# Regular Employee
regular_emp = Employee(
    name="John", age=28, gender="Male", department="Engineering", salary=75000
)
```

```python
# Manager
store_manager = Manager(
    name="Peter",
    age=38,
    gender="Male",
    department="Operations",
    salary=90000,
    team_size=12,
)
```

```python
print("--- Employee JSON Record ---")
# json.dumps converts the dictionary to a nicely formatted JSON string
print(json.dumps(regular_emp.get_details(), indent=4))
```

```python
print("\n--- Manager JSON Record ---")
print(json.dumps(store_manager.get_details(), indent=4))
```
